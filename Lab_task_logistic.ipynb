```python
import pandas as pd
```

```python
train_df = pd.read_csv("/content/train.csv", encoding="latin1")
test_df = pd.read_csv("/content/test.csv", encoding="latin1")
```

```python
print(train_df["sentiment"].value_counts())
```

## Remove Neutral Sentiments


```python
train_df = train_df[
     train_df["sentiment"].isin(["positive", "negative"])
]

test_df = test_df[
    test_df["sentiment"].isin(["positive", "negative"])
]
```

```python
print(train_df["sentiment"].value_counts())
print(test_df["sentiment"].value_counts())
```

```python
train_df = train_df.dropna(subset=["text", "sentiment"])
test_df = test_df.dropna(subset=["text", "sentiment"])
```

seperate input and output


```python
X_train = train_df["text"]
y_train = train_df["sentiment"]

X_test = test_df["text"]
y_test = test_df["sentiment"]
```

 Convert Text into Numerical Features

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=5000
)

X_train_new = vectorizer.fit_transform(X_train)

X_test_new = vectorizer.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(X_train_new, y_train)
```

```python
y_pred = model.predict(X_test_new)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

TESTING


```python
new_text = ["I really love this movie. It is amazing!"]
```

```python
new_text_vector = vectorizer.transform(new_text)
```

```python
prediction = model.predict(new_text_vector)

print(prediction)
```
